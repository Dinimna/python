```python
#afficher l'ensemble des clés d'un dictionnaire#
mon_dictionnaire={
    "marie" :15,
    "landry" :14,
    "eric" :13,
    "samuel" :10,
    "sandrine" :12,
}
print(mon_dictionnaire.keys())
```

```python
#afficher l'ensemble des valeurs d'un dictionnaire#
mon_dictionnaire.values()
```

```python
#parcourir les clés et les valeurs d'un dictionnaire avec la boucle for#
for cle in mon_dictionnaire.keys():
    print(cle)
```

```python
for valeur in mon_dictionnaire.keys():
    print(valeur)
```

```python
#parcourir simultanement les clé et les valeur#
mon_dictionnaire.items()
```

```python
for cle,valeur in mon_dictionnaire.items():
    print(cle,valeur)
```
